```python
import pandas as pd 
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
```

```python
data = pd.read_csv("detect_dataset.csv")
```

```python
data.head()
```

```python
data1 = data.drop("Unnamed: 7", axis =1)
```

```python
data1.head()
```

```python
data2 = data1.drop("Unnamed: 8", axis =1)
```

```python
data2.head()
```

```python
X = data2.drop("Output (S)", axis =1 )
Y = data2["Output (S)"] 
```

```python
X
```

```python
Y
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2)
```

```python
X_train
```

```python
Y_train
```

```python
rfc = RandomForestClassifier(n_estimators = 20)
rfc.fit(X_train, Y_train)
```

```python
Y_pred = rfc.predict(X_test)
```

```python
Y_pred
```

```python
accuracy_score(Y_test, Y_pred )
```

```python
cols = data2.columns.drop("Output (S)")
feature_importance = pd.Series(rfc.feature_importances_, index = cols)
feature_importance
```
